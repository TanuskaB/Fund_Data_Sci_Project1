### Step 2: Cleaning up our datasets ###

```python
#creating a range of years from 2000 to 2024 for ATL Ozone dataset to match the AQI dataset using a masking label of
#True/False values to filter the dataset
mask_atl = (df_atl['Date'] >= '2000-01-01') & (df_atl['Date'] <= '2024-12-31')
df_atl = df_atl.loc[mask_atl]

mask_la = (df_la['Date'] >= '2000-01-01') & (df_la['Date'] <= '2024-12-31')
df_la = df_la.loc[mask_la]

#removing null/unknown columns from AQI, ATL Ozone, and LA Ozone datasets
df_aqi = df_aqi.drop(columns= 'Main Pollutant')
df_atl = df_atl.drop(columns= 'Main Pollutant')
df_la = df_la.drop(columns= 'Main Pollutant')

#changing the data type of the 'Date' column from a string to a datetime
df_aqi['Date'] = pd.to_datetime(df_aqi['Date'])
df_atl['Date'] = pd.to_datetime(df_atl['Date'])
df_la['Date'] = pd.to_datetime(df_la['Date'])


```

### Step 3: Merging Datasets (AQI vs. ATL Ozone) and (ATL Ozone vs LA Ozone) ###

```python
df_aqi_combined = pd.merge(df_aqi[['Date', 'PM2.5 AQI Value']], df_atl[['Date','Ozone AQI Value']], on='Date', how= 'inner')
print(df_aqi_combined.sample(10))

#renaming the columns Ozone AQI Value for ATL and LA so they are easy to identify when merged together
atl_ozone = df_atl[['Date', 'Ozone AQI Value']].rename(columns={'Ozone AQI Value': 'ATL_Ozone'})
la_ozone = df_la[['Date', 'Ozone AQI Value']].rename(columns={'Ozone AQI Value': 'LA_Ozone'})

#merging the two cloumns together with their column in common which is Date
df_cities_combined = pd.merge(atl_ozone, la_ozone, on= 'Date', how= 'inner')
#df_cities_combined.head(5)
print(df_cities_combined.sample(10))
```

### Step 4: Correlation of the datasets ###

```python
correlation_aqi_atl = df_aqi_combined[['PM2.5 AQI Value', 'Ozone AQI Value']].corr()
print(correlation_aqi_atl)

correlation_cities = df_cities_combined[['ATL_Ozone', 'LA_Ozone']].corr()
print(correlation_cities)
```

### Step 5: Data Visualizations ###

#### PM2.5 AND Ozone for ATL ####

##### 1. Yearly Trends of ATL: #####

```python
df_aqi_combined['Year'] = df_aqi_combined['Date'].dt.year

#taking the day and grouping it by year and the finding the avg of both the PM2.5 AQI Value and Ozone AQI Value for ATL
yearly_atl = df_aqi_combined.groupby('Year')[['PM2.5 AQI Value', 'Ozone AQI Value']].mean()

plt.figure(figsize=(10, 5))
plt.plot(yearly_atl.index, yearly_atl['PM2.5 AQI Value'], label='ATL PM2.5 (Avg)', marker='o', color='blue',)
plt.plot(yearly_atl.index, yearly_atl['Ozone AQI Value'], label='ATL Ozone (Avg)', marker='s', color='orange')
plt.title('Atlanta Air Quality Improvement (2000-2024)')
plt.ylabel('Average AQI')
plt.legend()
plt.show()
```

##### 2. Proving Ozone relies on weather conditions "Summer Spikes" while PM2.5 does not: #####

```python
df_aqi_combined['Month'] = df_aqi_combined['Date'].dt.month

#taking the day and grouping it by month and the finding the avg of both the PM2.5 AQI Value and Ozone AQI Value for ATL
monthly_atl = df_aqi_combined.groupby('Month')[['PM2.5 AQI Value', 'Ozone AQI Value']].mean()

plt.figure(figsize=(10, 5))
plt.bar(monthly_atl.index, monthly_atl['Ozone AQI Value'], color='orange', alpha=0.6, label='ATL Ozone (Avg)')
plt.plot(monthly_atl.index, monthly_atl['PM2.5 AQI Value'], color='blue', marker='o', label='ATL PM2.5 (Avg)')
plt.title('Pollution Patterns by Month in Atlanta')
plt.xticks(range(1, 13)) #represents the months
plt.legend()
plt.show()
```

#### ATL vs LA Comparisons ####

##### 1. ATL Ozone vs LA Ozone Levels #####

```python
#taking the day and grouping it by year and the finding the avg of both Ozone Levels for ATL and LA
df_cities_combined['Year'] = df_cities_combined['Date'].dt.year
city_yearly_avg = df_cities_combined.groupby('Year')[['ATL_Ozone', 'LA_Ozone']].mean()

# creating the visualization
plt.figure(figsize=(12, 6))

# plotting ATL trend
plt.plot(city_yearly_avg.index, city_yearly_avg['ATL_Ozone'], 
         label='ATL Ozone', marker='o', color='red', linewidth=2)

# plotting LA trend
plt.plot(city_yearly_avg.index, city_yearly_avg['LA_Ozone'], 
         label='LA Ozone', marker='o', color='steelblue', linewidth=2)

# formatting
plt.title('Ozone Trend Comparison: Atlanta vs. Los Angeles (2000-2024)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Ozone AQI Value', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
```

##### 2. Showing the ATL vs LA Ozone Levels and the AQI Threshold: #####

```python
#set a clean visual style
sns.set_theme(style="whitegrid")

#create the boxplot comparing the distribution of Ozone in both cities
plt.figure(figsize=(10, 6))
sns.boxplot(data=df_cities_combined[['ATL_Ozone', 'LA_Ozone']], palette="Set1")

#adding labels and title to make it professional
plt.title('Ozone Intensity Distribution: ATL vs. LA (2000-2024)', fontsize=14)
plt.ylabel('Ozone AQI Value', fontsize=12)
plt.xlabel('Cities', fontsize=12)

#added a dashed line at 100 to show the "Unhealthy" threshold
plt.axhline(y=100, color='blue', linestyle='--', label='Unhealthy Threshold (100 AQI)')
plt.legend()

plt.show()
```
