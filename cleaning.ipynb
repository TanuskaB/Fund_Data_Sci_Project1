```python
import pandas as pd
import numpy as np

# Load raw dataset
df = pd.read_csv('Dataset for DS Project - ATL_ALL_AQI_2000_2025.csv')
print("Original Shape:", df.shape)
print(df.dtypes)
df.head()

# Convert Date column to datetime and drop invalid dates
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y', errors='coerce')
df = df.dropna(subset=['Date'])

# Clean string columns by trimming whitespace
string_cols = df.select_dtypes(include=['object']).columns
df[string_cols] = df[string_cols].apply(lambda col: col.str.strip())

# Normalize missing Main Pollutant values
df['Main Pollutant'] = df['Main Pollutant'].fillna('Unknown').astype(str)

# Ensure PM2.5 AQI values are numeric and remove invalid rows
df['PM2.5 AQI Value'] = pd.to_numeric(df['PM2.5 AQI Value'], errors='coerce')
df = df.dropna(subset=['PM2.5 AQI Value'])

# Remove duplicate rows and sort dataset
original_rows = df.shape[0]
df = df.drop_duplicates().sort_values('Date').reset_index(drop=True)
print(f'Dropped {original_rows - df.shape[0]} duplicate or invalid rows')

# Save cleaned dataset
output_path = 'Dataset for DS Project - ATL_ALL_AQI_2000_2025_cleaned.csv'
df.to_csv(output_path, index=False)
print('Cleaned Shape:', df.shape)
print('Saved cleaned dataset to', output_path)
df.head()
```
